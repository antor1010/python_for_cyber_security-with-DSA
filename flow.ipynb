```python
#positive or negetive even odd
num = int(input("Enter interger number: "))

if num > 0:
    print("Positive", end=" ")
    if num % 2 == 0:
        print("Even")
    else:
        print("Odd")

elif num < 0:
    print("Negative", end=" ")
    if num % 2 == 0:
        print("Even")
    else:
        print("Odd")

else:
    print("Zero")
```
