```python
#days after
days = ["Sunday", "Monday", "Tuesday",
        "Wednesday", "Thursday", "Friday", "Saturday"]
a=int(input("day:"))
n=int(input("after:"))
res=(a+n)%7
print(days[res])
```

```python
#before
days = ["Sunday", "Monday", "Tuesday",
        "Wednesday", "Thursday", "Friday", "Saturday"]
a=int(input("day:"))
n=int(input("before:"))
res=(a-n)%7
print(days[res])
```
